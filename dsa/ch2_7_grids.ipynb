```python
def test_ex_2_7_7():

    def board(*cells):
        b = [[0] * 9 for _ in range(9)]
        for r, c, d in cells:
            b[r][c] = d
        return b
    full = [[(r * 3 + r // 3 + c) % 9 + 1 for c in range(9)] for r in range(9)]
    assert valid_board(board())
    assert valid_board(full)
    assert valid_board(board((0, 0, 5), (4, 4, 5)))
    assert not valid_board(board((0, 0, 5), (1, 1, 5)))
    assert not valid_board(board((0, 4, 3), (8, 4, 3)))
    assert not valid_board(board((4, 0, 7), (4, 8, 7)))
    full[8][8] = full[8][7]
    assert not valid_board(full)


test_ex_2_7_7()
print('2.7.7 passed')
```
