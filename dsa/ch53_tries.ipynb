```python
# Support code used by the exercises: run it as is.
class WordNode:
    """Trie node for 53.10-53.16: `next` maps a letter to a child, `passes` counts the words through it, `ends` the words ending at it."""

    def __init__(self):
        self.next = {}
        self.passes = 0
        self.ends = 0
```

```python
# Support code used by the exercises: run it as is.
def word_trie(words):
    """Build a `WordNode` trie holding `words`, repeats counted."""
    root = WordNode()
    for word in words:
        node = root
        node.passes += 1
        for ch in word:
            node = node.next.setdefault(ch, WordNode())
            node.passes += 1
        node.ends += 1
    return root
```
